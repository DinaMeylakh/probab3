```python
import numpy as np
from scipy import interpolate
from matplotlib import pyplot as plt

file_data = np.loadtxt('spera_mapelli_initial_final_BH.csv',delimiter=',',skiprows=1)
mass_zams = file_data[:,0]
mass_rem = file_data[:,-1]
initial_final_func = interpolate.interp1d(mass_zams, mass_rem)
```

```python
# Plot interplotation
xs = np.linspace(19.5, 111, 100)
ys = initial_final_func(xs)

plt.plot(mass_zams, mass_rem, '.')
plt.plot(xs, ys)
plt.xlabel("$M_{ZAMS}$")
plt.ylabel("$M_{rem}$")
plt.show()
```

```python
# Save interpolation data to file
import pickle

with open('initial_final_interp.pck', 'wb') as interp_file:
    pickle.dump(initial_final_func, interp_file)
```

```python
# Load interpolation data from file
import pickle

with open('initial_final_interp.pck', 'rb') as interp_file:
    loaded_initial_final_func = pickle.load(interp_file)
```

```python
# Show that loading works
xs = np.linspace(25, 110, 100)
ys = loaded_initial_final_func(xs)

plt.plot(xs, ys)
plt.xlabel("$M_{ZAMS}$")
plt.ylabel("$M_{rem}$")
plt.show()
```
